```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense, Conv2D, AveragePooling2D, Flatten
```

```python
from keras.datasets import mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

# LeNet Architecture

```python
# Initialize the Sequential model [2]
model = Sequential()



# Layer 1: Convolutional Layer [2-4]
# 6 filters, kernel size 5x5, stride 1, no padding, activation 'tanh'
model.add(Conv2D(6, kernel_size=(5,5), padding='valid',
                 activation='tanh', input_shape=(32,32,1)))
# Layer 2: Average Pooling Layer [2-4]
# Pool size 2x2, stride 2
model.add(AveragePooling2D(pool_size=(2,2), strides=2, padding='valid'))



# Layer 3: Convolutional Layer [2-4]
# 16 filters, kernel size 5x5, activation 'tanh'
model.add(Conv2D(16, kernel_size=(5,5), padding='valid', activation='tanh'))
# Layer 4: Average Pooling Layer [2-4]
# Pool size 2x2, stride 2
model.add(AveragePooling2D(pool_size=(2,2), strides=2, padding='valid'))



# Flattening the 3D output to 1D [2, 4, 5]
model.add(Flatten())



# Layer 5: Fully Connected Layer [2, 3, 6]
# 120 neurons, activation 'tanh'
model.add(Dense(120, activation='tanh'))
# Layer 6: Fully Connected Layer [2, 3, 6]
# 84 neurons, activation 'tanh'
model.add(Dense(84, activation='tanh'))
# Output Layer [2, 5, 6]
# 10 neurons (for 10 digits), activation 'softmax'
model.add(Dense(10, activation='softmax'))
```
